```python
import pandas as pd
import csv

file_path = "/home/thant_syn/exp/polar/processed/task2_polarization_type_no_id.txt"

columns = [
    "text",
    "political",
    "racial/ethnic",
    "religious",
    "gender/sexual",
    "other"
]

rows = []

with open(file_path, "r", encoding="utf-8-sig", newline="") as f:
    for line_no, line in enumerate(f, start=1):
        line = line.rstrip("\r\n")

        if not line:
            continue

        # Skip the header
        if line_no == 1 and line.startswith("text|political|"):
            continue

        # Split from the right to preserve pipes in the text
        parts = line.rsplit("|", 5)

        if len(parts) != 6:
            print(f"Skipping malformed line {line_no}")
            continue

        rows.append(parts)

df2 = pd.DataFrame(rows, columns=columns)

# Convert labels to numeric values
label_cols = columns[1:]
df2[label_cols] = df2[label_cols].apply(
    pd.to_numeric, errors="coerce"
)

print("Dataset shape:", df2.shape)
display(df2.head())
print("\nMissing values:")
print(df2.isna().sum())
```

### Split the dataset

We'll use an 80% training, 10% validation, and 10% test split.

```python
from iterstrat.ml_stratifiers import MultilabelStratifiedShuffleSplit

split1 = MultilabelStratifiedShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, temp_idx = next(split1.split(X, y))

X_train, X_temp = X[train_idx], X[temp_idx]
y_train, y_temp = y[train_idx], y[temp_idx]

split2 = MultilabelStratifiedShuffleSplit(
    n_splits=1,
    test_size=0.50,
    random_state=42
)

val_idx, test_idx = next(split2.split(X_temp, y_temp))

X_val, X_test = X_temp[val_idx], X_temp[test_idx]
y_val, y_test = y_temp[val_idx], y_temp[test_idx]

print("Training:", X_train.shape, y_train.shape)
print("Validation:", X_val.shape, y_val.shape)
print("Testing:", X_test.shape, y_test.shape)
```

### Tokenize and pad Burmese text

```python
import tensorflow as tf
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Hyperparameters
MAX_WORDS = 20000
MAX_LEN = 100
EMBEDDING_DIM = 128

# Fit tokenizer on training data only
tokenizer = Tokenizer(
    num_words=MAX_WORDS,
    oov_token="<OOV>"
)
tokenizer.fit_on_texts(X_train)

# Convert text to sequences
X_train_seq = tokenizer.texts_to_sequences(X_train)
X_val_seq = tokenizer.texts_to_sequences(X_val)
X_test_seq = tokenizer.texts_to_sequences(X_test)

# Pad sequences
X_train_pad = pad_sequences(
    X_train_seq, maxlen=MAX_LEN, padding="post", truncating="post"
)
X_val_pad = pad_sequences(
    X_val_seq, maxlen=MAX_LEN, padding="post", truncating="post"
)
X_test_pad = pad_sequences(
    X_test_seq, maxlen=MAX_LEN, padding="post", truncating="post"
)

vocab_size = min(MAX_WORDS, len(tokenizer.word_index) + 1)

print("Vocabulary size:", vocab_size)
print("Training shape:", X_train_pad.shape)
print("Validation shape:", X_val_pad.shape)
print("Test shape:", X_test_pad.shape)
```

## **Build the BiLSTM model**

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Embedding, Bidirectional, LSTM, Dense, Dropout
)
from tensorflow.keras.callbacks import EarlyStopping

model = Sequential([
    Embedding(
        input_dim=vocab_size,
        output_dim=EMBEDDING_DIM,
        mask_zero=True
    ),
    Bidirectional(LSTM(64, return_sequences=True)),
    Dropout(0.3),
    Bidirectional(LSTM(32)),
    Dropout(0.3),
    Dense(32, activation="relu"),
    Dropout(0.2),
    Dense(5, activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=[
        tf.keras.metrics.BinaryAccuracy(name="binary_accuracy")
    ]
)

model.summary()
```

```python
early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

history = model.fit(
    X_train_pad,
    y_train,
    validation_data=(X_val_pad, y_val),
    epochs=15,
    batch_size=32,
    callbacks=[early_stopping],
    verbose=1
)
```

```python
## evaluate model

from sklearn.metrics import classification_report, f1_score, hamming_loss

y_prob = model.predict(X_test_pad)
y_pred = (y_prob >= 0.5).astype(int)

print("Macro F1:", f1_score(y_test, y_pred, average="macro", zero_division=0))
print("Micro F1:", f1_score(y_test, y_pred, average="micro", zero_division=0))
print("Hamming Loss:", hamming_loss(y_test, y_pred))

print("\nPer-category report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=label_cols,
    zero_division=0
))
```

```python
import re

import emoji

def clean_text(text):
    # Remove emojis
    text = emoji.replace_emoji(text, replace="")

    # Normalize multiple spaces, tabs, and other whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text


input_file = "task2_text_raw.txt"
output_file = "task2_text_clean.txt"

with open(input_file, "r", encoding="utf-8") as fin:
    lines = fin.readlines()

with open(output_file, "w", encoding="utf-8") as fout:
    for line in lines:
        cleaned = clean_text(line)
        fout.write(cleaned + "\n")

print(f"Cleaning complete: {output_file}")
```

```python
import numpy as np
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.multiclass import OneVsRestClassifier
from sklearn.svm import LinearSVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    f1_score,
    classification_report,
    accuracy_score,
)

TEXT_FILE = "task2_text_word.txt"
LABEL_FILE = "task2_labels.csv"

LABEL_COLS = [
    "political",
    "racial/ethnic",
    "religious",
    "gender/sexual",
    "other",
]

# Load data
texts = pd.read_csv(
    TEXT_FILE, header=None, dtype=str, keep_default_na=False
)[0].tolist()

Y = pd.read_csv(
    LABEL_FILE, header=None
).to_numpy(dtype=int)

# Validate alignment
if len(texts) != len(Y):
    raise ValueError(
        f"Text/label mismatch: {len(texts)} texts vs {len(Y)} labels"
    )

if Y.shape[1] != len(LABEL_COLS):
    raise ValueError(f"Expected {len(LABEL_COLS)} label columns, got {Y.shape[1]}")

# Keep rows aligned; do not independently drop blank text lines
X = np.array(texts, dtype=object)

# Use the same random seed and test size as your baseline
indices = np.arange(len(X))
train_idx, test_idx = train_test_split(
    indices,
    test_size=0.2,
    random_state=42,
)

X_train, X_test = X[train_idx].tolist(), X[test_idx].tolist()
y_train, y_test = Y[train_idx], Y[test_idx]

# OPPA has already segmented the text.
# Split on whitespace to use the supplied word boundaries.
pipeline = TfidfVectorizer(
    analyzer="word",
    tokenizer=str.split,
    preprocessor=None,
    token_pattern=None,
    lowercase=False,
    ngram_range=(1, 2),
    sublinear_tf=True,
)

X_train_vec = pipeline.fit_transform(X_train)
X_test_vec = pipeline.transform(X_test)

model = OneVsRestClassifier(
    LinearSVC(C=1.0)
)
model.fit(X_train_vec, y_train)

predictions = model.predict(X_test_vec)

print("Accuracy (subset accuracy):",
      round(accuracy_score(y_test, predictions), 4))
print("Macro F1 (labels):",
      round(f1_score(y_test, predictions, average="macro",
                     zero_division=0), 4))
print("Micro F1:",
      round(f1_score(y_test, predictions, average="micro",
                     zero_division=0), 4))

print("\nPer-label report:")
print(classification_report(
    y_test,
    predictions,
    target_names=LABEL_COLS,
    zero_division=0,
))

print("\nPer-label F1:")
for label, score in zip(
    LABEL_COLS,
    f1_score(y_test, predictions, average=None, zero_division=0),
):
    print(f"{label}: {score:.4f}")
```
